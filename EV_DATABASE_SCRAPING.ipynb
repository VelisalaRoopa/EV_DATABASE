```python
import pandas as pd
import numpy as np
import seaborn as sns
import requests
import re
from bs4 import BeautifulSoup
import warnings
warnings.filterwarnings('ignore')
```

```python
url = 'https://ev-database.org/#group=vehicle-group&rs-pr=10000_100000&rs-er=0_1000&rs-ld=0_1000&rs-ac=2_23&rs-dcfc=0_400&rs-ub=10_200&rs-tw=0_3000&rs-ef=100_350&rs-sa=-1_5&rs-w=1000_3500&rs-c=0_5000&rs-y=2010_2030&s=1&p=0-10'
```

```python
response = requests.get(url)
```

```python
response
```

```python
pagecontent = response.text
```

```python
pagecontent
```

```python
soup = BeautifulSoup(pagecontent)
```

```python
productname = []

for x in soup.find_all('a', class_='title'):
    model = x.find('span', class_='hidden')
    if model:
        productname.append(model.text.strip())

for name in productname:
    print(name)
len(productname)
```

```python
availability = []

for x in soup.find_all("div", class_="availability"):
    availability.append(x.get_text(strip=True))

for name in availability:
    print(name)

print(len(availability))
```

```python
erange = []

for x in soup.find_all("span", class_="erange_real"):
    erange.append(x.get_text(strip=True))

print(erange)
print(len(erange))
```

```python
efficiency = []

for x in soup.find_all("span", class_="efficiency"):
    efficiency.append(x.get_text(strip=True))

print(efficiency)
print(len(efficiency))
```

```python
weight = []

for x in soup.find_all("span", class_="weight_p"):
    weight.append(x.get_text(strip=True))

print(weight)
print(len(weight))
```

```python
accelerate = []

for x in soup.find_all("span", class_="acceleration_p"):
    accelerate.append(x.get_text(strip=True))

print(accelerate)
print(len(accelerate))
```

```python
highway_speed = []

for x in soup.find_all("span", class_="long_distance_total"):
    highway_speed.append(x.get_text(strip=True))

print(highway_speed)
print(len(highway_speed))
```

```python
battery = []

for x in soup.find_all("span", class_="battery_p"):
    battery.append(x.get_text(strip=True))

print(battery)
print(len(battery))
```

```python
Fastcharge = []

for x in soup.find_all("span", class_="fastcharge_speed_print"):
    Fastcharge.append(x.get_text(strip=True))

print(Fastcharge)
print(len(Fastcharge))
```

```python
Towing = []

for x in soup.find_all("span", class_="towweight_p"):
    Towing.append(x.get_text(strip=True))

print(Towing)
print(len(Towing))
```

```python
Cargo_vol = []

for x in soup.find_all("span", class_="cargo"):
    Cargo_vol.append(x.get_text(strip=True))

print(Cargo_vol)
print(len(Cargo_vol))
```

```python
Price = []

for x in soup.find_all("span", class_="priceperrange_p"):
    Price.append(x.get_text(strip=True))

print(Price)
print(len(Price))
```

```python
price_de_list = []

for item in soup.find_all("div", class_="item-data"):
    
    # Find Germany price inside this car block
    price_de_tag = item.find("span", class_="country_de")
    
    if price_de_tag:
        price_de_list.append(price_de_tag.text.strip())
print(price_de_list)
print(len(price_de_list))
```

```python
price_nl_list = []

for item in soup.find_all("div", class_="item-data"):
    
    # Find Germany price inside this car block
    price_nl_tag = item.find("span", class_="country_nl")
    
    if price_nl_tag:
        price_nl_list.append(price_nl_tag.text.strip())
print(price_nl_list)
print(len(price_nl_list))
```

```python
price_uk_list = []

for item in soup.find_all("div", class_="item-data"):
    
    # Find Germany price inside this car block
    price_uk_tag = item.find("span", class_="country_uk")
    
    if price_uk_tag:
        price_uk_list.append(price_uk_tag.text.strip())
print(price_uk_list)
print(len(price_uk_list))
```

```python
drive_type_list = []

for item in soup.find_all("div", class_="item-data"):
    
    # Find drive icon span using data-tooltip
    drive_span = item.find("span", attrs={"data-tooltip": True})
    
    if drive_span:
        drive_type = drive_span["data-tooltip"]
        drive_type_list.append(drive_type)
print(drive_type_list)
print(len(drive_type_list))
```

```python
seats_list = []

for item in soup.find_all("div", class_="item-data"):
    
    icons_section = item.find("div", class_="icons")
    
    seats = np.nan
    
    if icons_section:
        seat_icon = icons_section.find("i", class_=re.compile(r"seats-\d+"))
        
        if seat_icon:
            classes = seat_icon["class"]
            
            for cls in classes:
                if cls.startswith("seats-"):
                    seats = int(cls.split("-")[1])
    
    seats_list.append(seats)
print(seats_list)
print(len(seats_list))
```

```python
heatpump_list = []

for item in soup.find_all("div", class_="item-data"):
    
    icons_section = item.find("div", class_="icons")
    
    heatpump_available = 0   # default = Not Available
    
    if icons_section:
        heatpump_icon = icons_section.find(
            "span", 
            attrs={"data-tooltip": "Heat pump available"}
        )
        
        if heatpump_icon:
            heatpump_available = 1
    
    heatpump_list.append(heatpump_available)
print(heatpump_list)
print(len(heatpump_list))
```

```python
towbar_list = []

for item in soup.find_all("div", class_="item-data"):
    
    icons_section = item.find("div", class_="icons")
    
    towbar_available = 0  # default = No
    
    if icons_section:
        towbar_icon = icons_section.find(
            "span",
            attrs={"data-tooltip": "Towbar possible"}
        )
        
        if towbar_icon:
            towbar_available = 1
    
    towbar_list.append(towbar_available)

print(towbar_list)
print(len(towbar_list))
```

```python
safety_rating_list = []

for item in soup.find_all("div", class_="item-data"):
    
    safety_tag = item.find("span", class_="safety")
    
    if safety_tag:
        rating = int(safety_tag.text.strip())
        
        # Convert -1 to NaN
        if rating == -1:
            rating = np.nan
            
        safety_rating_list.append(rating)
    else:
        safety_rating_list.append(np.nan)

print(safety_rating_list)
print(len(safety_rating_list))
```

```python
bidirectional_list = []
v2l_list = []
v2h_list = []
v2g_list = []

for item in soup.find_all("div", class_="item-data"):
    
    # Default values
    bidirectional = 0
    v2l = 0
    v2h = 0
    v2g = 0
    
    # Check each tooltip
    spans = item.find_all("span", attrs={"data-tooltip": True})
    
    for span in spans:
        tooltip = span["data-tooltip"]
        
        if "Bi-directional charging possible" in tooltip:
            bidirectional = 1
            
        if "Vehicle-2-Load" in tooltip and "possible" in tooltip:
            v2l = 1
            
        if "Vehicle-2-Home" in tooltip and "possible" in tooltip:
            v2h = 1
            
        if "Vehicle-2-Grid" in tooltip and "possible" in tooltip:
            v2g = 1
    
    bidirectional_list.append(bidirectional)
    v2l_list.append(v2l)
    v2h_list.append(v2h)
    v2g_list.append(v2g)

print(bidirectional_list)
print(len(bidirectional_list))
print(v2l_list)
print(len(v2l_list))
print(v2h_list)
print(len(v2h_list))
print(v2g_list)
print(len(v2g_list))
```

```python
df = pd.DataFrame({'product_name':productname, 'Availability': availability, 'Price': Price, 'Vechile_range':erange, 'Efficiency':efficiency, 'Weight':weight,'Acceleration':accelerate, 'Long_distance':highway_speed, 'battery': battery, 'Fast_Charge': Fastcharge, 'Towing': Towing, 'Cargo_vol': Cargo_vol, 'German_Price' : price_de_list, 'Netherlands_Price': price_nl_list, 'Uk_Price': price_uk_list, 'Driver_Type': drive_type_list, 'No.of_Seats': seats_list, 'Heat_Pump': heatpump_list, 'Towbar': towbar_list, 'Rating': safety_rating_list, 'Bidirectional': bidirectional_list, 'Load': v2l_list, 'Home': v2h_list, 'Grid': v2g_list})
```
